# 2. Tratamento de ruído (Filtragem Espacial – Suavização)
Deverá ser introduzido artificialmente um nível controlado de ruído em cada imagem original. O
grupo deverá escolher entre filtro Gaussiano, filtro da mediana ou outra técnica de suavização
devidamente justificada. A intensidade do ruído utilizada deverá ser informada e mantida controlada para
permitir a comparação entre os resultados.

- Filtro escolhido: Filtro da mediana com janela 3×3.
    - Justificativa: Será adicionado artificialmente ruído do tipo sal e pimenta em 5% dos pixels das imagens. O filtro da mediana foi escolhido por apresentar bom desempenho na remoção de ruídos impulsivos, pois substitui cada pixel pela mediana dos valores presentes em sua vizinhança. Dessa forma, valores extremos introduzidos pelo ruído possuem menor influência no resultado. Foi utilizada inicialmente uma janela 3×3 por proporcionar redução do ruído com menor perda de detalhes e contornos importantes da imagem.
    
    - **Ruído artificial:** Sal e pimenta.
    - **Intensidade:** 5% dos pixels.
    - **Janela do filtro:** 3×3.
    - **Implementação:** As operações serão implementadas manualmente, sem utilização de funções prontas de filtragem.

```python
def adicionar_ruido_sal_pimenta(imagem, percentual=0.05, seed=42):
    """
    Adiciona ruído do tipo sal e pimenta em uma imagem.

    percentual:
        Fração dos pixels que serão alterados.
        0.05 corresponde a 5%.

    seed:
        Mantém a geração do ruído reproduzível.
    """

    # Criamos uma cópia para não alterar a imagem original
    imagem_ruidosa = imagem.copy()

    # Gerador de números aleatórios
    rng = np.random.default_rng(seed)

    # Altura e largura da imagem
    altura, largura = imagem.shape[:2]

    # Quantidade total de pixels
    total_pixels = altura * largura

    # Calcula quantos pixels correspondem ao percentual escolhido
    quantidade_ruido = int(total_pixels * percentual)

    # Seleciona posições aleatórias sem repetir pixels
    indices = rng.choice(
        total_pixels,
        size=quantidade_ruido,
        replace=False
    )

    # Divide os pixels entre sal e pimenta
    metade = quantidade_ruido // 2

    indices_pimenta = indices[:metade]
    indices_sal = indices[metade:]

    # Converte índice linear em linha e coluna
    linhas_pimenta = indices_pimenta // largura
    colunas_pimenta = indices_pimenta % largura

    linhas_sal = indices_sal // largura
    colunas_sal = indices_sal % largura

    # Aplica pimenta = preto
    imagem_ruidosa[linhas_pimenta, colunas_pimenta] = 0

    # Aplica sal = branco
    imagem_ruidosa[linhas_sal, colunas_sal] = 255

    return imagem_ruidosa
```

```python
def filtro_mediana_3x3(imagem):
    """
    Aplica filtro da mediana 3x3 sem utilizar
    funções prontas de filtragem.

    Para cada pixel, são obtidos os 9 valores
    da vizinhança 3x3. Esses valores são ordenados
    e o elemento central é utilizado como resultado.
    """

    # Mantém uma cópia da imagem original.
    # As bordas permanecerão inalteradas.
    imagem_filtrada = imagem.copy()

    # Cria todas as vizinhanças 3x3 da imagem.
    # Nenhum filtro pronto é aplicado aqui:
    # apenas organizamos os pixels em janelas.
    janelas = np.lib.stride_tricks.sliding_window_view(
        imagem,
        window_shape=(3, 3),
        axis=(0, 1)
    )

    # Formato:
    # altura-2 x largura-2 x canais x 3 x 3
    altura_janelas, largura_janelas, canais = janelas.shape[:3]

    # Transforma cada janela 3x3 em uma sequência de 9 valores.
    janelas = janelas.reshape(
        altura_janelas,
        largura_janelas,
        canais,
        9
    )

    # Ordena os 9 valores de cada vizinhança.
    valores_ordenados = np.sort(janelas, axis=-1)

    # Em uma lista de 9 valores ordenados,
    # a mediana está na posição 4.
    medianas = valores_ordenados[..., 4]

    # Insere as medianas na região interna da imagem.
    imagem_filtrada[1:-1, 1:-1, :] = medianas

    return imagem_filtrada
```

```python
"""APLICAÇÃO DO TRATAMENTO DE RUÍDO"""

# As quatro imagens já foram carregadas no início do notebook
imagens = [
    ("ISIC_0011410", ISIC1),
    ("ISIC_0011865", ISIC2),
    ("ISIC_0024396", ISIC3),
    ("ISIC_0024994", ISIC4)
]

# Lista que armazenará os resultados desta etapa
resultados_ruido = []

for indice, (nome, imagem_original) in enumerate(imagens):

    print(f"Processando {nome}...")

    # Adiciona ruído sal e pimenta em 5% dos pixels
    imagem_ruidosa = adicionar_ruido_sal_pimenta(
        imagem_original,
        percentual=0.05,
        seed=42 + indice
    )

    # Aplica o filtro da mediana 3x3 implementado manualmente
    imagem_filtrada = filtro_mediana_3x3(imagem_ruidosa)

    # Guarda as três versões da imagem para as próximas etapas
    resultados_ruido.append({
        "nome": nome,
        "original": imagem_original,
        "ruidosa": imagem_ruidosa,
        "filtrada": imagem_filtrada
    })

print("Tratamento de ruído concluído.")
```

```python
"""COMPARAÇÃO DOS RESULTADOS"""

plt.figure(figsize=(18, 20))

for i, resultado in enumerate(resultados_ruido):

    # Conversão BGR -> RGB apenas para exibição no Matplotlib
    original_rgb = resultado["original"][:, :, ::-1]
    ruidosa_rgb = resultado["ruidosa"][:, :, ::-1]
    filtrada_rgb = resultado["filtrada"][:, :, ::-1]

    # Imagem original
    plt.subplot(4, 3, i * 3 + 1)
    plt.imshow(original_rgb)
    plt.title(f'{resultado["nome"]} - Original')
    plt.axis("off")

    # Imagem com ruído
    plt.subplot(4, 3, i * 3 + 2)
    plt.imshow(ruidosa_rgb)
    plt.title("Ruído sal e pimenta - 5%")
    plt.axis("off")

    # Imagem após a mediana
    plt.subplot(4, 3, i * 3 + 3)
    plt.imshow(filtrada_rgb)
    plt.title("Mediana 3x3")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

### Análise dos resultados

A aplicação do ruído sal e pimenta introduziu pixels com valores extremos
nas quatro imagens. Como o nível de ruído foi mantido em 5%, foi possível
realizar o processamento de forma controlada e reproduzível.

Após a aplicação do filtro da mediana com janela 3×3, grande parte dos
pixels afetados pelo ruído foi removida. Isso ocorre porque o filtro utiliza
o valor mediano da vizinhança, reduzindo a influência dos valores extremos
0 e 255 introduzidos pelo ruído sal e pimenta.

Visualmente, o filtro preservou as principais estruturas e contornos das
lesões, embora seja possível observar uma pequena suavização de detalhes
finos. Esse comportamento demonstra o compromisso entre redução de ruído
e preservação das informações relevantes da imagem.
