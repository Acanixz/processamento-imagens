# Descrição do projeto a ser desenvolvido
O processamento digital de imagens possui papel relevante em aplicações computacionais
voltadas à saúde, principalmente quando imagens médicas precisam ser preparadas para análise
humana ou para etapas posteriores de classificação e reconhecimento automático. Em diferentes bases
de imagens, variações de iluminação, contraste, ruído e definição das estruturas podem interferir na
interpretação dos dados e no desempenho de métodos computacionais.

O conjunto ISIC é composto por câncer de pele e permite trabalhar com imagens de diferentes
regiões da pele. Para este trabalho, o grupo deverá selecionar utilizar de três a quatro imagens
representativas, priorizando imagens que apresentem diferenças perceptíveis de contraste, ruído ou
definição estrutural.

Neste trabalho, o foco será o desenvolvimento de um pipeline de melhoria de imagens médicas
utilizando operações pontuais e técnicas de filtragem espacial. Embora o conjunto de dados utilizado
continue sendo o International Skin Imaging Collaboration ([ISIC](https://www.kaggle.com/datasets/nodoubttome/skin-cancer9-classesisic)), cada grupo deverá trabalhar com uma sequência de imagens diferente daquela utilizada em trabalhos de outros grupos e justificar a escolha

O objetivo não é simplesmente aplicar filtros, mas investigar como diferentes transformações
alteram as características visuais da imagem e determinar, por meio de métricas quantitativas, qual
combinação apresenta o melhor compromisso entre melhoria do contraste, redução de ruído e
preservação ou realce de detalhes.

# Desafios no Processamento de Imagens Médicas
1. Variação de contraste: diferenças de intensidade podem dificultar a visualização de estruturas e
regiões de interesse.
2. Presença de ruído: interferências podem mascarar informações relevantes e prejudicar
operações posteriores.
3. Perda de detalhes: filtros de suavização podem eliminar pequenas estruturas juntamente com o
ruído.
4. Realce excessivo: operações de aguçamento podem produzir saturações ou intensificar ruídos
e artefatos.
5. Diferentes características entre imagens: uma técnica que apresenta bom resultado em uma
imagem pode não ser a mais adequada para outra, exigindo análise e comparação.
