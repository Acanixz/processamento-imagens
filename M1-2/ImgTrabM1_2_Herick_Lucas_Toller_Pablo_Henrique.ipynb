# 1. Transformação de intensidade (Operação Pontual)
O grupo deverá escolher uma técnica de transformação de intensidade entre equalização de
histograma, correção gama ou outra operação pontual tecnicamente justificada. A escolha deverá
considerar as características das imagens selecionadas. Além de apresentar o resultado visual, o grupo
deverá explicar por que a técnica escolhida é adequada ao problema observado e comparar seu efeito
nas três imagens.


- Técnica escolhida: Correção gama (γ) automática, com γ calculado a partir do brilho médio de cada imagem
    - Justificativa: As quatro imagens possuem níveis de contraste muito diferentes entre si (ISIC1 e ISIC4 com contraste baixo/baixíssimo, enquanto ISIC2 e ISIC3 já possuem melhor distribuição de intensidade). Uma transformação fixa aplicada igualmente às quatro, como a equalização de histograma global, correria o risco de sobre-realçar as imagens que já têm boa distribuição de intensidade.
    A correção gama, por ser uma transformação suave e monotônica (s = c·r^γ), evita esse problema: escolhendo γ automaticamente a partir do brilho médio de cada imagem (γ = ln(0.5) / ln(média_normalizada).
    Por fim, aplicamos o mesmo γ igualmente nos três canais de cor (B, G, R), o que preserva a relação de cor entre eles — importante em imagens de dermatoscopia, já que a coloração/pigmentação da lesão é uma informação diagnosticamente relevante que não deve ser distorcida pela etapa de realce de contraste.

```python
"""1. TRANSFORMAÇÃO DE INTENSIDADE (OPERAÇÃO PONTUAL) - CORREÇÃO GAMA ADAPTATIVA"""

def luminancia(img_bgr):
    """Converte BGR para luminância (escala de cinza) usando a fórmula ponderada
    padrão (0.299R + 0.587G + 0.114B), usada aqui para medir o brilho médio."""
    b, g, r = cv2.split(img_bgr.astype(np.float64))
    y = 0.299 * r + 0.587 * g + 0.114 * b
    return np.clip(y, 0, 255).astype(np.uint8)

def calcular_gamma_automatico(canal_cinza):
    """Calcula o gamma que leva o brilho médio da imagem para o centro da escala (127.5),
    Imagens já centradas (media ~127) resultam em gamma ~1 (quase nenhuma alteração)."""
    media = canal_cinza.mean() / 255.0
    media = np.clip(media, 1e-6, 1 - 1e-6)  # evita log 0 e div. 0 em imagens totalmente pretas/brancas
    gamma = np.log(0.5) / np.log(media)
    return np.clip(gamma, 0.2, 5.0)  # evita gammas extremos em casos degenerados

def aplicar_gama(imagem, gamma):
    """Aplica a transformação de potência s = 255 * (r/255)^gamma via LUT (0-255),
    igualmente nos 3 canais de cor para preservar a relação de cor entre eles."""
    i = np.arange(256, dtype=np.float64)
    lut = np.clip(255.0 * (i / 255.0) ** gamma, 0, 255).astype(np.uint8)
    return lut[imagem]

def corrigir_intensidade(img_bgr):
    """Pipeline da Parte 1: calcula o gamma automático a partir do brilho médio
    (luminância) da imagem e aplica a correção gama na imagem colorida original."""
    gamma = calcular_gamma_automatico(luminancia(img_bgr))
    return aplicar_gama(img_bgr, gamma), gamma
```

```python
# Teste do pipeline: Correção de Gama adaptativa

imagens_originais = [ISIC1, ISIC2, ISIC3, ISIC4]
nomes = ['ISIC1 (0011410)', 'ISIC2 (0011865)', 'ISIC3 (0024396)', 'ISIC4 (0024994)']

resultados_gama = [corrigir_intensidade(img) for img in imagens_originais]
imagens_corrigidas = [r[0] for r in resultados_gama]
gammas = [r[1] for r in resultados_gama]

# ---------- Comparação visual: original x corrigida ----------
fig, ax = plt.subplots(2, 4, figsize=(20, 9))
for c, (orig, corr, nome, gamma) in enumerate(zip(imagens_originais, imagens_corrigidas, nomes, gammas)):
    ax[0, c].imshow(orig[:, :, ::-1])  # BGR -> RGB apenas para exibição
    ax[0, c].set_title(f'{nome}\nOriginal (média={luminancia(orig).mean():.1f})')
    ax[0, c].axis('off')

    ax[1, c].imshow(corr[:, :, ::-1])
    ax[1, c].set_title(f'Corrigida (γ={gamma:.2f})\nmédia={luminancia(corr).mean():.1f}')
    ax[1, c].axis('off')
plt.tight_layout()
plt.show()

# ---------- Comparação quantitativa: desvio padrão e faixa dinâmica (proxy de contraste) ----------
print("Comparativo de contraste (luminância) antes -> depois da correção gama:")
for nome, orig, corr, gamma in zip(nomes, imagens_originais, imagens_corrigidas, gammas):
    y_orig, y_corr = luminancia(orig), luminancia(corr)
    print(f"  {nome:20s} | gamma={gamma:5.2f} | "
          f"std {y_orig.std():5.1f} -> {y_corr.std():5.1f} | "
          f"faixa [{y_orig.min():3d},{y_orig.max():3d}] -> [{y_corr.min():3d},{y_corr.max():3d}]")
```
